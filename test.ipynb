```python
from logic import *

digits = "0123456789"
positions = [0, 1, 2]

# créer tous les symboles digit-position
symbols = {
    (d, p): Symbol(f"d{d}_p{p}")
    for d in digits
    for p in positions
}

knowledge = And()

# chaque position contient exactement un chiffre
for p in positions:
    knowledge.add(
        Or(*[symbols[(d, p)] for d in digits])
    )

    for d1 in digits:
        for d2 in digits:
            if d1 != d2:
                knowledge.add(
                    Implication(symbols[(d1, p)], Not(symbols[(d2, p)]))
                )

# ---------- CLUES ----------

# 530 : aucun chiffre correct
for d, p in zip("530", positions):
    knowledge.add(Not(symbols[(d, p)]))

# 548 : un chiffre correct et bien placé
knowledge.add(
    Or(
        symbols[("5",0)],
        symbols[("4",1)],
        symbols[("8",2)]
    )
)

# 157 : deux chiffres corrects mais mauvaise place
knowledge.add(
    And(
        Or(symbols[("1",1)], symbols[("1",2)]),
        Or(symbols[("5",0)], symbols[("5",2)]),
        Or(symbols[("7",0)], symbols[("7",1)])
    )
)

# 806 : un chiffre correct mais mauvaise place
knowledge.add(
    Or(
        symbols[("8",1)], symbols[("8",2)],
        symbols[("0",0)], symbols[("0",2)],
        symbols[("6",0)], symbols[("6",1)]
    )
)

# 647 : un chiffre correct mais mauvaise place
knowledge.add(
    Or(
        symbols[("6",1)], symbols[("6",2)],
        symbols[("4",0)], symbols[("4",2)],
        symbols[("7",0)], symbols[("7",1)]
    )
)

# ---------- chercher la solution ----------

solution = ["?","?","?"]

for d in digits:
    for p in positions:
        if model_check(knowledge, symbols[(d,p)]):
            solution[p] = d

print("PIN:", "".join(solution))
```
